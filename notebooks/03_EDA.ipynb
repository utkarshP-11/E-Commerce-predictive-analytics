```python
import pandas as pd

df = pd.read_csv(
    "/content/drive/MyDrive/Colab Notebooks/E-Commerce Customer Intelligence & Sales Analytics/data/retail_features.csv",
    parse_dates=['InvoiceDate']
)
```

```python
df['Revenue'].sum()             # total revenue
```

```python
df['CustomerID'].nunique()            # total customers
```

```python
df['InvoiceNo'].nunique()           # total orders
```

```python
df.groupby('InvoiceNo')['Revenue'].sum().mean()           # average order value
```

```python
monthly = df.groupby('Month')['Revenue'].sum()

monthly.plot(kind='line')
```

```python
country = (
    df.groupby('Country')['Revenue']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

country.plot(kind='bar')
```

```python
products = (
    df.groupby('Description')['Revenue']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

products.plot(kind='bar')
```

```python
hourly = df.groupby('Hour')['Revenue'].sum()

hourly.plot(kind='line')
```
