```python
import pandas as pd

df = pd.read_csv(
    "/content/drive/MyDrive/Colab Notebooks/E-Commerce Customer Intelligence & Sales Analytics/data/cleaned_retail.csv",
    parse_dates=['InvoiceDate']
)
```

```python
df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
df['MonthName'] = df['InvoiceDate'].dt.month_name()
df['Day'] = df['InvoiceDate'].dt.day
df['Hour'] = df['InvoiceDate'].dt.hour
df['Weekday'] = df['InvoiceDate'].dt.day_name()
```

```python
df[['InvoiceDate','Year','MonthName','Weekday','Hour']].head()
```

```python
order_revenue = (
    df.groupby('InvoiceNo')['Revenue']
      .sum()
      .reset_index()
)

order_revenue.head()
```
