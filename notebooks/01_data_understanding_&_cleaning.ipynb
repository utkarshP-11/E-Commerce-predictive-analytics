```python
df = df.dropna(subset=['CustomerID'])                   #Removing rows with missing Description
```

```python
df = df.dropna(subset=['Description'])                  #Removing rows with missing Description
```

```python
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]         # Remove cancelled invoices
```

```python
df = df[df['Quantity'] > 0]                      # Remove Quantity ≤ 0
```

```python
df = df[df['UnitPrice'] > 0]                      # Remove UnitPrice ≤ 0
```

```python
df['Revenue'] = df['Quantity'] * df['UnitPrice']
df[['Quantity','UnitPrice','Revenue']].head()
```
